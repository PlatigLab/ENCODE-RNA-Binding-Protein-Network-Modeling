```python
# Load in packages
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import polars as pl

_=pl.Config.set_tbl_cols(100000)
_=pl.Config.set_tbl_rows(10000)
_=pl.Config.set_tbl_width_chars(10000)
_=pl.Config.set_fmt_str_lengths(10000)
```

```python
path = "/project/PlatigLab/users/yogi/ENCODE-RNA-Binding-Protein-Network-Modeling/analysis/07_second_order_SHAP_analysis/2_analyze_shap_interaction_values/outputs/SHAP_cache/averages/Signed-Local-SHAP-Mean-Bound-Only.tsv"
```

```python
table = pd.read_csv(path, sep='\t')
```

```python
# Keep rows where either RBP1 or RBP2 equals "SAFB"
df_filtered = table[(table['RBP 1'] == 'SAFB') | (table['RBP 2'] == 'SAFB')]
```

```python
# Sort values high-low
df_filtered = df_filtered.sort_values(by='Value - Signed-Local-SHAP-Mean-Bound-Only', ascending=False)
df_filtered.to_csv('SAFB_value_sorted.csv', index=False)

# Remove 0s
df_filtered = df_filtered[df_filtered['Value - Signed-Local-SHAP-Mean-Bound-Only'].notna() & (df_filtered['Value - Signed-Local-SHAP-Mean-Bound-Only'] != 0)]
df_filtered.to_csv('SAFB_no_zeros.csv', index=False)

# Get top values based on magnitude
df_filtered['Value - Signed-Local-SHAP-Mean-Bound-Only'] = df_filtered['Value - Signed-Local-SHAP-Mean-Bound-Only'].abs()
mag = df_filtered.sort_values(by='Value - Signed-Local-SHAP-Mean-Bound-Only', ascending=False)
mag.to_csv('SAFB_magnitude.csv', index=False)
```

```python
interactions = table[
    ((table['RBP 1'] == 'IGF2BP1') & (table['RBP 2'] == 'PTBP1')) |
    ((table['RBP 1'] == 'PTBP1') & (table['RBP 2'] == 'IGF2BP1'))
]
```
