```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
```

```python
df = pd.read_csv(f'/project/PlatigLab/users/reece/ENCODE-RNA-Binding-Protein-Network-Modeling/analysis/07_reece_biological_validations/2_rbp_pair_shap_correlations/data/final_K562_matrix_ppis.csv')
```

```python
# Extract the Pearson_Corr column as a Series
pearson = df["Pearson_Corr"].dropna()

sns.set(style="whitegrid")

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Distribution
sns.histplot(pearson, kde=False, bins=20, ax=axes[0], color="skyblue")
axes[0].set_title("Distribution of Pearson Correlations", fontsize=14)
axes[0].set_xlabel("Pearson_Corr")
axes[0].set_ylabel("Frequency")
axes[0].set_ylim(0, 15)

# Probability Density Function
sns.kdeplot(pearson, ax=axes[1], fill=True, color="teal")
axes[1].set_title("Probability Density Function (PDF)", fontsize=14)
axes[1].set_xlabel("Pearson_Corr")
axes[1].set_ylabel("Density")

# Cumulative Density Function
sns.kdeplot(pearson, ax=axes[2], fill=True, cumulative=True, color="purple")
axes[2].set_title("Cumulative Density Function (CDF)", fontsize=14)
axes[2].set_xlabel("Pearson_Corr")
axes[2].set_ylabel("Cumulative Probability")

plt.tight_layout()
plt.show()
```

```python
mean = pearson.mean()
```
