### Get Reference Files 


```python

reference_files = {}

# reference_files["hg38"] = pandas.read_csv("/home/jve4pt/resource-files/genes/hg38_gencode_v41_UCSC_browser.bed", sep="\t", header=None)[[0,1,2,3]]
reference_files["hg38"] = pandas.read_csv("/home/jve4pt/tmp.bed", sep="\t", header=None)

reference_files["hg38"][0] = "chr"+ reference_files["hg38"][0].astype(str)

reference_files
```

```python


reference_files["hg38"] = pybedtools.BedTool.from_dataframe(reference_files["hg38"])
reference_files["hg38"].head()
```

## Intersect Ranges

```python
bed_files["ENCFF315HYU"].intersect(reference_files["hg38"]).head()
```

```python
annotated_dataframes = {}

# for each bed file 
# intersect bed file with hg38 reference file 
# annotate each intersected region 
# return as Pandas dataframe 
for accession_id in bed_files: 
    accession_id
    annotated_dataframes[accession_id] = bed_files[accession_id].intersect(
        reference_files["hg38"], 
        wa=True, 
        wb=True, 
        loj=True
    ).to_dataframe(header=None) 
```
