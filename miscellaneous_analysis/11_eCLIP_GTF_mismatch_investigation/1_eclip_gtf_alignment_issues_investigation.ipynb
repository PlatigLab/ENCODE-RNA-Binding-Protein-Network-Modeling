# Per `cell line`, check the number of `RBPs` that are unique to `V29`, those with both `V24` and `V29`, and those with only `V24 eCLIP`

```python
import pandas as pd
pd.set_option('display.max_rows', 1000000)
pd.set_option('display.max_columns', 1000000)
pd.set_option('display.max_colwidth', 10000000)

from IPython.core.interactiveshell import InteractiveShell
InteractiveShell.ast_node_interactivity = "all"
```

```python
eclip_metadata = pd.read_csv(
    "../../inputs/metadata/eclip_metadata.tsv", 
    sep="\t" 
)

eclip_metadata.head()
```

```python
filtered = eclip_metadata[
    (eclip_metadata["File assembly"] == "GRCh38") & 
    (eclip_metadata["Biological replicate(s)"] == "1, 2") &
    (eclip_metadata["File Status"] == "released")
]

filtered["Genome annotation"].value_counts(dropna=False)
```

```python
# Get unique cell lines
cell_lines = filtered["Biosample term name"].unique()

for cell_line in sorted(cell_lines):
    # Filter data for this cell line
    cell_line_data = filtered[filtered["Biosample term name"] == cell_line]
    
    # Get unique experiment targets for V29 annotations
    v29_targets_cl = set(cell_line_data[cell_line_data["Genome annotation"] == "V29"]["Experiment target"])
    
    # Get unique experiment targets for V24 (NaN) genome annotations
    nan_targets_cl = set(cell_line_data[cell_line_data["Genome annotation"].isna()]["Experiment target"])
    
    # Check if V29 targets are a subset of NaN targets
    is_subset_cl = v29_targets_cl.issubset(nan_targets_cl)
    
    # Find which V29 RBPs are NOT in the V24 set
    not_in_nan_cl = v29_targets_cl - nan_targets_cl
    
    # Find which V29 RBPs ARE in the V24 set (intersection)
    in_both_cl = v29_targets_cl & nan_targets_cl
    
    print(f"\nCell line: {cell_line}")
    print(f"  V29 RBPs are a complete subset of V24 RBPs: {is_subset_cl}")
    print(f"  Number of RBPs in V29: {len(v29_targets_cl)}")
    print(f"  Number of RBPs in V24: {len(nan_targets_cl)}")
    print(f"  Number of RBPs found in BOTH V29 and V24: {len(in_both_cl)}")
    
    if not_in_nan_cl:
        print(f"  V29 RBPs NOT found in V24 set: {len(not_in_nan_cl)}")
        for target in sorted(not_in_nan_cl):
            print(f"    - {target}")
    else:
        print(f"  All V29 RBPs are available in the V24 set!")

    if in_both_cl:
        print(f"  V29 RBPs found in BOTH sets: {len(in_both_cl)}")
        for target in sorted(in_both_cl):
            print(f"    - {target}")
    else:
        print(f"  No V29 RBPs are found in the V24 set.")
    
```
